### Demographic (DEMO) dataset overview

The 'DEMO' dataset provides the demographic information, such as age, sex, race, education, poverty status, family status and other background characteristics. 

There are 27 variables. Let's briefly review each of them.
1. SEQN - Respondent sequence number, we can use SEQN to connect with other tables.
2. SDDSRVYR - Data release cycle. A value of '12' means August 2021-August 2023 data. Our whole dataset is based on this cycle.
3. RIDSTATR - Interview/Examination status. 
   - Status '1' identifies a participant was only interviewed but never participated in the examination.
   - Status '2' identifies a participant was both interviewed and examined in the mobile examination center (MEC).
4. RIAGENDR - Gender.
   - '1' is a male.
   - '2' is a female.
5. RIDAGEYR - Age in years at the time of screening. Individuals 80 and over are top-coded at 80 years of age.
6. RIDAGEMN - Age in months at the time of screening. Field used only for children aged 24 months or younger at the time of exam (or screening if not examined).
7. RIDRETH1 (old parameter, to link with previous studies) - Race-ethnicity categories 
   - '1' - '4' - variables are the same as `RIDRETH3`
   - '5' - used only in RIDRETH1 for Asian participants and  other non-Hispanic races.
8. RIDRETH3 - Race-ethnicity categories
   -  '1' - respondents who self-identified as "Mexican American".
   -  '2' - participants who self-identified as "Hispanic".
   -  '3' - individuals who self-reported their race as "non-Hispanic White".
   -  '4' - those who self-reported their race as "non-Hispanic Black".
   -  '6' - those who self-reported their race as "non-Hispanic Asian".
   -  '7' - other non-Hispanic races, including non-Hispanic multiracial.
9. RIDEXMON - Examination period. Six month time period when the examination was performed.
   - '1' - November 1 through April 30.
   - '2' - May 1 through October 31.
10. RIDEXAGM - Age in months at the time of examination for people aged 19 years or younger.
11. DMQMILIZ - Whether the respondent has ever served active duty in US Armed Forces. 
   - '1' - Yes
   - '2' - No
   - '7' - Refused (to answer)
   - '9' - Don't know
12. DMDBORN4 -  Country of birth. 
   - '1' - Born in 50 U.S. states or Washington, DC.
   - '2' - Born in other countries, including U.S. territories
   - '77' - Refused (to answer)
   - '99' - Don't know
13. DMDYRUSR - Length of time living in the US.
   - '1' - Less than 1 year
   - '2' - 1 to 4 years
   - '3' - 5 to 9 years
   - '4' - 10 to 14 years
   - '5' - 15 to 19 years
   - '6' - 20 years or more
   - '77' - Refused (to answer)
   - '99' - Don't know
14. DMDEDUC2 - The highest level of education the respondent has achieved. Collected for people above 20.
   - '1' - Less than 9th grade
   - '2' - 9-11th grade (Includes 12th grade with no diploma)
   - '3' - High school graduate/GED or equivalent
   - '4' - Some college or AA degree
   - '5' - College graduate or above
   - '7' - Refused (to answer)
   - '9' - Don't know
15. DMDMARTZ - Marital status. Collected for people above 20.
   - '1' - Married/Living with partner
   - '2' - Widowed/Divorced/Separated
   - '3' - Never married
   - '77' - Refused (to answer)
   - '99' - Don't know
16. RIDEXPRG - Pregnancy status at exam for females only 20 - 44 years.
   - '1' - Yes, positive lab pregnancy test or self-reported pregnant at exam
   - '2' - The participant was not pregnant at exam
   - '3' - Cannot ascertain if the participant is pregnant at exam
17. DMDHHSIZ -  Total number of people in the household.
   - '1' - '6' - range from 1 to 6 people in the household
   - '7' - 7 or more people in the household

   The next variables with the `DMDHR` prefix contain the background information of the household reference person (parent or legal guardian) for youth participants aged 0–19:
   
18. DMDHRGND - Household Reference Person’s gender.
   - '1' - Male.
   - '2' - Female.
19. DMDHRAGZ - Household Reference Person’s age in years.
   - '1' - <20 years
   - '2' - 20-39 years
   - '3' - 40-59 years
   - '4' - 60+ years
20. DMDHREDZ - Household Reference Person’s education level.
   - '1' - Less than high school degree
   - '2' - High school grad/GED or some college/AA degree
   - '3' - College graduate or above
21. DMDHRMAZ - Household Reference Person’s marital status.
   - '1' - Married/Living with partner
   - '2' - Widowed/Divorced/Separated
   - '3' - Never Married
22. DMDHSEDZ - Household Reference Person’s spouse’s education level
   - '1' - Less than high school degree
   - '2' - High school grad/GED or some college/AA degree
   - '3' - College graduate or above
23. WTINT2YR - Full sample 2-year interview weight. 
24. WTMEC2YR - Full sample 2-year MEC exam weight. The value is 0 in case the individual has never been examined.

*Note: The 2-year sample weights (`WTINT2YR`, `WTMEC2YR`, depending on the analysis) should be used for all NHANES August 2021-August 2023 analyses to estimate characteristics of the entire US population. **However, this exploratory analysis uses unweighted data. Therefore, the results should be interpreted as describing the NHANES study sample rather than the entire US population.***

25. SDMVSTRA - Masked variance pseudo-stratum.
26. SDMVPSU - Masked variance pseudo-PSU.

*Note: Masked Variance Units (`SDMVSTRA`, `SDMVPSU`) are variables describing the complex NHANES sampling design. They are required for survey-weighted statistical analysis **but are not used in the current exploratory analysis.***

27. INDFMPIR - Ratio of family income to the federal poverty threshold. Values below 1 indicate that the family's income is below the poverty line.

### Mental health - depression screener (DPQ) dataset overview

The 'DPQ' dataset contains the responses the Patient Health Questionnaire (also, PHQ-9), which provides the frequency of depression symptoms over the past two weeks. There are 9 questions for each depression symptom, the participants scored how frequently they had experienced those symptoms recently from 0 to 3. In addition, there was 1 question (`DPQ100`) about how much those symptoms bothered the participant.

A total score, ranging from 0 to 27, can be calculated for participants who have complete answers to the symptom questions (`DPQ010` - `DPQ090`). Major depression and depression severity can be assessed using pre-defined cut-points, described in the references.

There are 11 variables. Let's briefly review each of them.

1. SEQN - Respondent sequence number, we can use SEQN to connect with other tables.
2. DPQ010 - Have a little interest in doing things.
    
    _The question is "Over the last 2 weeks, how often have you been bothered by the following problems: little interest or pleasure in doing things? Would you say..."_

3. DPQ020  - Feeling down, depressed, or hopeless.

    _The question is "Over the last 2 weeks, how often have you been bothered by the following problems: feeling down, depressed, or hopeless?"_

4. DPQ030  - Trouble sleeping or sleeping too much.

    _The question is "Over the last 2 weeks, how often have you been bothered by the following problems: trouble falling or staying asleep, or sleeping too much?"_

5. DPQ040  - Feeling tired or having little energy.

    _The question is "Over the last 2 weeks, how often have you been bothered by the following problems: feeling tired or having little energy?"_

6. DPQ050 - Poor appetite or overeating.
    
    _The question is "Over the last 2 weeks, how often have you been bothered by the following problems: poor appetite or overeating?"_

7. DPQ060 - Feeling bad about yourself.

    _The question is "Over the last 2 weeks, how often have you been bothered by the following problems: feeling bad about yourself - or that you are a failure or have let yourself or your family down?"_

8. DPQ070 - Trouble concentrating on things.

    _The question is "Over the last 2 weeks, how often have you been bothered by the following problems: trouble concentrating on things, such as reading the newspaper or watching TV?"_

9. DPQ080 - Moving or speaking slowly or too fast.

    _The question is "Over the last 2 weeks, how often have you been bothered by the following problems: moving or speaking so slowly that other people could have noticed? Or the opposite - being so fidgety or restless that you have been moving around a lot more than usual?"_

10. DPQ090 - Thought you would be better off dead.

    _The question is "Over the last 2 weeks, how often have you been bothered by the following problems: thoughts that you would be better off dead or of hurting yourself in some way?"_

    **The variables `DPQ010` - `DPQ090` may have the same possible values:**
    - '0' - Not at all
    - '1' - Several days
    - '2' - More than half the days	
    - '3' - Nearly every day
    - '7' - Refused (to answer)
    - '9' - Don't know

11. DPQ100 - Difficulty these problems have caused.
    
    _The question is "How difficult have these problems made it for you to do your work, take care of things at home, or get along with people?"_

    **The variable has the possible values:**
    - '0' - Not at all difficult
    - '1' - Somewhat difficult
    - '2' - Very difficult
    - '3' - Extremely difficult
    - '7' - Refused (to answer)
    - '9' - Don't know

### Preparing the DPQ dataset

To calculate PHQ-9 total score, which is required to estimate depression levels, we need to select only valid answers to the first 9 symptom questions (`DPQ010` - `DPQ090`). Their values range from 0 to 3. However, before filtering participants, we need to clean and check the imported XPT values.


#### 1. Inspecting response values 

```python
dpq.sample(10)
```

```python
dpq_columns = ['DPQ010', 'DPQ020', 'DPQ030', 'DPQ040','DPQ050','DPQ060','DPQ070', 'DPQ080', 'DPQ090', 'DPQ100']
for column in dpq_columns:
    print(dpq[column].value_counts(dropna=False).sort_index(), end='\n\n')
  
```

During the inspection, we observe a discrepancy between the NHANES documentation and the data. According to the NHANES codebook, participants could respond with '0', which means 'not at all'/'not at all difficult'. 

Nevertheless, we observe a '5' answer, which is not documented. 
The reason for that discrepancy might happen during importing the XPT files into the pandas library. 

The responses that should represent '0' were imported from the XPT file as extremely small positive floating-point values. These values do not correspond to any valid PHQ-9 response category and are treated as zeros.

#### 2. Cleaning imported 'zero' values

```python
dpq_clean = dpq.copy()

dpq_clean[dpq_columns] = dpq_clean[dpq_columns].mask(
    dpq_clean[dpq_columns] < 1e-70,
    0
)
```

```python
#Check the result

for column in dpq_columns:
    print(dpq_clean[column].value_counts(dropna=False).sort_index(), end='\n\n')
  
```

#### 3. Selecting records with valid answers to symptom questions

```python
phq9_columns = ['DPQ010', 'DPQ020', 'DPQ030', 'DPQ040','DPQ050','DPQ060','DPQ070', 'DPQ080', 'DPQ090']

valid_phq9 = dpq_clean[phq9_columns].isin([0, 1, 2, 3]).all(axis=1)

filtered_dpq = dpq_clean[valid_phq9].copy()

filtered_dpq
```

```python
for column in dpq_columns:
    print(filtered_dpq[column].value_counts(dropna=False).sort_index(), end='\n\n')
```

#### 2. Investigating missing values

##### DPQ100 
`DPQ100` measures how much depressive symptoms interfere with daily life. Unlike the first nine questions, it is not included in the PHQ-9 total score.

```python
dpq10_missing = analysis_df[analysis_df['DPQ100'].isnull()]
dpq10_missing
```

```python
phq9_columns = ['DPQ010', 'DPQ020', 'DPQ030', 'DPQ040','DPQ050','DPQ060','DPQ070', 'DPQ080', 'DPQ090']
phq9_all_zero = dpq10_missing[
    (dpq10_missing[phq9_columns] == 0).all(axis=1)
]
phq9_all_zero
```

```python
len(phq9_all_zero) == len(dpq10_missing)
```

```python
not_all_zero = dpq10_missing[
    ~(dpq10_missing[phq9_columns] == 0).all(axis=1)
]

not_all_zero[phq9_columns + ['DPQ100'] + ['SEQN']]
```

```python
all_symptoms_zero = (
    dpq10_missing[phq9_columns]
    .eq(0)
    .all(axis=1)
)

print(f'Missing DPQ100: {len(dpq10_missing)}')
print(f'Structural missing values: {all_symptoms_zero.sum()}')
print(f'Other missing values: {(~all_symptoms_zero).sum()}')
```

##### DPQ100 summary
Of the 1292 participants with missing `DPQ100` values, **1291** reported 'Not at all' to all nine PHQ-9 symptom questions. Since `DPQ100` is asked only when at least one symptom is endorsed (according NHANES documentation), these _1291 missing values are structural_. One participant reported depressive symptoms but still had a missing `DPQ100` value, which appears to be a non-structural missing response. 

Since the participant (`SEQN = 139522.0`) has completed PHQ-9 responses, they are retained in the analytical dataset. However, this exaptional case will be taken into account in future analyses involving 'DPQ100'.

##### DMDYRUSR
`DMDYRUSR` reflects the length of time the participant has lived in the US. According to NHANES documentation, only participants born outside the US were asked the month and year when they came to live in the US. 
Let's verify that the missing values are structural.

```python
dmdyrusr_missing = analysis_df[analysis_df['DMDYRUSR'].isnull()]
dmdyrusr_missing
```

Since `DMDYRUSR` was collected only for participants who were born abroad, we can examine `DMDBORN4` variable, that indicates the participant's place of birth. `DMDBORN4` has two categories:
1. Born in 50 U.S. states or Washington, DC.
2. Born in other countries, including U.S. territories.

As expected, `DMDYRUSR` is missing for participants born in the US. Let's verify whether any participants classified as `DMDBORN4 = 2` (born in other countries or U.S. territories) also have missing `DMDYRUSR` values.

```python
dmdborn4_category2 = dmdyrusr_missing[dmdyrusr_missing['DMDBORN4']!=1]
dmdborn4_category2
```

##### DMDYRUSR summary
No participants with missing `DMDYRUSR` belonged to `DMDBORN4` category 2. Therefore, all missing `DMDYRUSR` values correspond to participants in `DMDBORN4` category 1, indicating that the missing values are structural rather than missing responses.

##### RIDEXPRG
`RIDEXPRG` represents pregnancy status. This data was released only for women aged 20-44 years. If a woman didn't know she was pregnant or didn't participate in the examination, the pregnancy status was assigned a value of '3', which means 'could not be determined'.
Thus, the structural missing data must be for men and women above 44 years old in our analysis dataset. 
Let's verify that the missing values are structural.

```python
ridexprg_missing = analysis_df[analysis_df['RIDEXPRG'].isnull()]
ridexprg_missing[['SEQN','RIAGENDR','RIDAGEYR','RIDEXPRG']]
```

```python
men_and_older_women = ridexprg_missing[(ridexprg_missing['RIAGENDR']==1) | (ridexprg_missing['RIDAGEYR']>44)]
men_and_older_women
```

```python
len(ridexprg_missing) == len(men_and_older_women)
```

##### RIDEXPRG summary
All missing values in `RIDEXPRG` are structural. There are no women aged 20-44 with missing `RIDEXPRG` values.


##### INDFMPIR
`INDFMPIR` represents the ratio of family income to poverty level. According to the NHANES documentation, `INDFMPIR` was not computed when detailed income information was unavailable or when income data were missing. Therefore, unlike the previous variables, the missing values in `INDFMPIR` are not fully explained by the questionnaire design and should be considered genuine missing values. Let's briefly look at missing data.

```python
indfmpir_missing = analysis_df[analysis_df['INDFMPIR'].isnull()]
indfmpir_missing
```

##### INDFMPIR summary
Participants with missing `INDFMPIR` values are retained in the analytical dataset because the current step focuses on data exploration rather than statistical analysis. The missing values will be addressed when analyzing the association between income and depressive symptoms.
